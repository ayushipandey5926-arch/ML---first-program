```python
import requests
import pandas as pd

url = "https://api.open-meteo.com/v1/forecast?latitude=22.7196&longitude=75.8577&current=temperature_2m,relative_humidity_2m,weather_code&timezone=Asia/Kolkata"

response = requests.get(url)
data = response.json()

time = data["current"]["time"]
timezone = data["timezone"]
humidity = data["current"]["relative_humidity_2m"]
temperature = data["current"]["temperature_2m"]

if temperature < 20:
  weather = "cool"
elif temperature < 30:
  weather = "warm and dry"
else:
  weather = "hot and humid"


df = pd.DataFrame({
    "Time": [time],
    "Timezone": [timezone],
    "Humidity": [humidity],
    "Temperature": [temperature],
    "Weather": [weather]
})

df.to_csv("weather.csv", index=False)
```
